## Plotting results
---

```python
import os
import numpy as np
import pandas as pd
pd.options.plotting.backend = "plotly"
from plotly_resampler import register_plotly_resampler
register_plotly_resampler(mode='auto')

from results_utils_feature_ablation import get_forecast_plot

root_directory = "/home/ubuntu/repos/TSPindorama/results_feature_ablation/"
```

```python
dataset_name = "ETTh1"
model = "LSTM"
features_of_interest = ['empty']
pred_len = 96
df_plot = get_forecast_plot(
    dataset_name=dataset_name,
    model=model,
    pred_len=pred_len,
    features_of_interest=features_of_interest,
    root_directory=root_directory,
    offset_idx=0
)

df_plot.plot(title=f"Forecast Plot for {dataset_name} - Model {model} - Features: {', '.join(features_of_interest)}",)
```
